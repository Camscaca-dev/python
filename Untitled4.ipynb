```python
from pprint import pprint

dados = []

while True:
  nome= input("nome do funcionario: ")
  numero= input("Numero do funcionario: ")
  salario = input("Salario do funcionario: ")
  salario = float(salario)

  descontoT = salario * 4 / 100

  if salario < 2000 or salario == 2000:
    desconto=0
    salario += desconto + descontoT

  if salario>2000 and salario <= 3500:
    desconto= salario*0.1
    salario += desconto + descontoT

  if salario>3500:
    desconto= salario*0.15
    salario += desconto + descontoT


  dados.append({"nome": nome, "numero": numero, "salario liquido": salario, "desconto imposto de renda": desconto})
  n = input("Deseja continuar? [s/n]")
  if n == "n":
    break

for dados in dados:
  pprint(dados)


```
