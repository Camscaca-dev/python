```
import pandas as pd

# Dados
vendas_dados = {
    'Setor': ['Eletrônicos', 'Moda', 'Eletrônicos', 'Moda', 'Alimentos', 'Alimentos'],
    'Vendas': [1500, 300, 2200, 450, 120, 80]
}

# Criando o DataFrame
df_vendas = pd.DataFrame(vendas_dados)

# Agrupando por setor e calculando soma e média
resultado = df_vendas.groupby('Setor')['Vendas'].agg(
    Soma_Total='sum',
    Media_Vendas='mean'
)

print(resultado)
```

```
import pandas as pd
import numpy as np

# Dados das pessoas
dados = {
    'Peso': [70, 85, 60, 95, 50],
    'Altura': [1.75, 1.80, 1.65, 1.70, 1.60]
}

# Criando o DataFrame
df = pd.DataFrame(dados)

# Calculando o IMC
df['IMC'] = df['Peso'] / (df['Altura'] ** 2)

# Criando a coluna Situação com np.where
df['Situação'] = np.where(df['IMC'] >= 25, 'Sobrepeso', 'Normal')

print(df)
```
