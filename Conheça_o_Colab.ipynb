```
menor = float('inf')
maior = 0

for i in range(15):
    altura = float(input("Digite a altura da pessoa: "))

    if altura < menor:
        menor = altura

    if altura > maior:
        maior = altura

print("A menor altura do grupo é:", menor)
print("A maior altura do grupo é:", maior)
```

```
soma = 0
quantidade = 0
positivos = 0
negativos = 0

numero = float(input("Digite um número (0 para parar): "))

while numero != 0:
    soma = soma + numero
    quantidade = quantidade + 1

    if numero > 0:
        positivos = positivos + 1
    else:
        negativos = negativos + 1

    numero = float(input("Digite um número (0 para parar): "))

media = soma / quantidade
perc_positivos = (positivos / quantidade) * 100
perc_negativos = (negativos / quantidade) * 100

print("Média:", media)
print("Quantidade de positivos:", positivos)
print("Quantidade de negativos:", negativos)
print("Percentual de positivos:", perc_positivos, "%")
print("Percentual de negativos:", perc_negativos, "%")
```

```
intervalo1 = 0
intervalo2 = 0
intervalo3 = 0
intervalo4 = 0

numero = float(input("Digite um número: "))

while numero >= 0:

    if numero >= 0 and numero <= 25:
        intervalo1 += 1

    elif numero >= 26 and numero <= 50:
        intervalo2 += 1

    elif numero >= 51 and numero <= 75:
        intervalo3 += 1

    elif numero >= 76 and numero <= 100:
        intervalo4 += 1

    numero = float(input("Digite um número: "))

print("Quantidade no intervalo [0-25]:", intervalo1)
print("Quantidade no intervalo [26-50]:", intervalo2)
print("Quantidade no intervalo [51-75]:", intervalo3)
print("Quantidade no intervalo [76-100]:", intervalo4)
```

```
soma = 0
soma_pares = 0
quantidade = 0
pares = 0
impares = 0

numero = int(input("Digite um número positivo (0 para parar): "))

while numero != 0:
    soma = soma + numero
    quantidade = quantidade + 1

    if numero % 2 == 0:
        pares = pares + 1
        soma_pares = soma_pares + numero
    else:
        impares = impares + 1

    numero = int(input("Digite um número positivo (0 para parar): "))

media_geral = soma / quantidade
media_pares = soma_pares / pares

print("Quantidade de números pares:", pares)
print("Quantidade de números ímpares:", impares)
print("Média dos valores pares:", media_pares)
print("Média geral:", media_geral)
```
