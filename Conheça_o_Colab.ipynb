```
# ==========================================
# Atividade 03 - Dataset Iris
# Importação, Filtros e Gráficos
# ==========================================

# Importando as bibliotecas
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Carregando o dataset Iris
iris = load_iris()

# Criando um DataFrame
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Adicionando o nome das espécies
df["species"] = [iris.target_names[i] for i in iris.target]

# Exibindo as primeiras linhas
print("Primeiras linhas do dataset:")
display(df.head())

# Informações gerais
print("\nInformações do Dataset:")
print(df.info())

print("\nEstatísticas:")
display(df.describe())

# ==================================================
# FILTRO 1
# Apenas flores da espécie Setosa
# ==================================================

setosa = df[df["species"] == "setosa"]

print("\nQuantidade de flores Setosa:", len(setosa))

plt.figure(figsize=(6,4))
sns.histplot(setosa["sepal length (cm)"], bins=10, color="blue")
plt.title("Comprimento da Sépala - Setosa")
plt.xlabel("Comprimento (cm)")
plt.ylabel("Quantidade")
plt.show()

# ==================================================
# FILTRO 2
# Flores com pétala maior que 5 cm
# ==================================================

petalas_grandes = df[df["petal length (cm)"] > 5]

print("\nQuantidade de flores com pétala > 5 cm:", len(petalas_grandes))

plt.figure(figsize=(6,4))
sns.countplot(data=petalas_grandes, x="species", palette="viridis")
plt.title("Espécies com Pétala maior que 5 cm")
plt.xlabel("Espécie")
plt.ylabel("Quantidade")
plt.show()

# ==================================================
# FILTRO 3
# Flores com largura da sépala maior que 3 cm
# ==================================================

sepala_larga = df[df["sepal width (cm)"] > 3]

print("\nQuantidade de flores com largura da sépala > 3 cm:", len(sepala_larga))

plt.figure(figsize=(7,5))
sns.scatterplot(
    data=sepala_larga,
    x="sepal length (cm)",
    y="petal length (cm)",
    hue="species",
    s=80
)

plt.title("Comprimento da Sépala x Comprimento da Pétala")
plt.xlabel("Comprimento da Sépala (cm)")
plt.ylabel("Comprimento da Pétala (cm)")
plt.show()
```
