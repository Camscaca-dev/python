```python
'''ex2'''
menor = float('inf')
maior = 0

for i in range(15):
    altura = float(input("Digite a altura da pessoa: "))

    if altura < menor:
        menor = altura

    if altura > maior:
        maior = altura

print("Menor altura:", menor)
print("Maior altura:", maior)
```

```python
'''ex3'''
soma = 0
quantidade = 0
positivos = 0
negativos = 0

numero = float(input("Digite um número (0 para parar): "))

while numero != 0:
    soma = soma + numero
    quantidade = quantidade + 1

    if numero > 0:
        positivos = positivos + 1
    else:
        negativos = negativos + 1

    numero = float(input("Digite um número (0 para parar): "))

media = soma / quantidade

perc_pos = (positivos / quantidade) * 100
perc_neg = (negativos / quantidade) * 100

print("Média:", media)
print("Quantidade de positivos:", positivos)
print("Quantidade de negativos:", negativos)
print("Percentual de positivos:", perc_pos, "%")
print("Percentual de negativos:", perc_neg, "%")
```

```python
'''ex5'''soma = 0
soma_pares = 0
quantidade = 0
pares = 0
impares = 0

numero = int(input("Digite um número (0 para parar): "))

while numero != 0:
    soma = soma + numero
    quantidade = quantidade + 1

    if numero % 2 == 0:
        pares = pares + 1
        soma_pares = soma_pares + numero
    else:
        impares = impares + 1

    numero = int(input("Digite um número (0 para parar): "))

media_geral = soma / quantidade

if pares > 0:
    media_pares = soma_pares / pares
else:
    media_pares = 0

print("Quantidade de números pares:", pares)
print("Quantidade de números ímpares:", impares)
print("Média dos valores pares:", media_pares)
print("Média geral:", media_geral)
```
