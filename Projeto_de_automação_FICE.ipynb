```python
import numpy as np
import pandas as pd
import math


# ==========================================
# FUNÇÕES MATEMÁTICAS BÁSICAS
# ==========================================

def operacoes_basicas():
    print("\n--- OPERAÇÕES BÁSICAS ---")

    a = float(input("Digite o primeiro número: "))
    b = float(input("Digite o segundo número: "))

    print("\nResultados:")
    print("Soma:", a + b)
    print("Subtração:", a - b)
    print("Multiplicação:", a * b)

    if b != 0:
        print("Divisão:", a / b)
    else:
        print("Divisão: não é possível dividir por zero")


# ==========================================
# EXPRESSÕES NUMÉRICAS
# ==========================================

def expressao_numerica():
    print("\n--- EXPRESSÃO NUMÉRICA ---")

    expressao = input("Digite uma expressão (ex: 10 + 5 * 2): ")

    try:
        # Permite apenas caracteres matemáticos básicos
        permitidos = "0123456789+-*/(). "

        if all(caractere in permitidos for caractere in expressao):
            resultado = eval(expressao)
            print("Resultado:", resultado)
        else:
            print("Expressão contém caracteres inválidos.")

    except:
        print("Não foi possível calcular a expressão.")


# ==========================================
# ESTATÍSTICA
# ==========================================

def estatistica():
    print("\n--- ESTATÍSTICA ---")

    entrada = input("Digite os números separados por espaço: ")

    numeros = [float(x) for x in entrada.split()]

    dados = pd.Series(numeros)

    print("\nDados:", list(dados))
    print("Média:", dados.mean())
    print("Mediana:", dados.median())
    print("Maior valor:", dados.max())
    print("Menor valor:", dados.min())
    print("Soma:", dados.sum())


# ==========================================
# MATRIZES
# ==========================================

def matrizes():
    print("\n--- MATRIZES ---")

    print("Digite os valores da matriz 2x2:")

    a = float(input("a11: "))
    b = float(input("a12: "))
    c = float(input("a21: "))
    d = float(input("a22: "))

    matriz = np.array([
        [a, b],
        [c, d]
    ])

    print("\nMatriz:")
    print(matriz)

    print("\nTransposta:")
    print(matriz.T)

    print("\nDeterminante:")
    print(np.linalg.det(matriz))


# ==========================================
# FUNÇÕES MATEMÁTICAS
# ==========================================

def funcoes_matematicas():
    print("\n--- FUNÇÕES MATEMÁTICAS ---")

    numero = float(input("Digite um número: "))

    print("\nResultados:")

    print("Raiz quadrada:", math.sqrt(numero))
    print("Quadrado:", numero ** 2)
    print("Cubo:", numero ** 3)

    if numero > 0:
        print("Logaritmo natural:", math.log(numero))
    else:
        print("Logaritmo: número inválido")


# ==========================================
# PROGRESSÃO ARITMÉTICA
# ==========================================

def progressao_aritmetica():
    print("\n--- PROGRESSÃO ARITMÉTICA ---")

    primeiro = float(input("Primeiro termo: "))
    razao = float(input("Razão: "))
    quantidade = int(input("Quantidade de termos: "))

    termos = []

    for i in range(quantidade):
        termo = primeiro + i * razao
        termos.append(termo)

    print("Progressão:")
    print(termos)

    print("Soma dos termos:", sum(termos))


# ==========================================
# GEOMETRIA
# ==========================================

def geometria():
    print("\n--- GEOMETRIA ---")

    print("1 - Área do quadrado")
    print("2 - Área do retângulo")
    print("3 - Área do círculo")
    print("4 - Área do triângulo")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        lado = float(input("Lado: "))
        area = lado ** 2
        print("Área:", area)

    elif opcao == "2":
        base = float(input("Base: "))
        altura = float(input("Altura: "))
        area = base * altura
        print("Área:", area)

    elif opcao == "3":
        raio = float(input("Raio: "))
        area = math.pi * raio ** 2
        print("Área:", area)

    elif opcao == "4":
        base = float(input("Base: "))
        altura = float(input("Altura: "))
        area = (base * altura) / 2
        print("Área:", area)

    else:
        print("Opção inválida.")


# ==========================================
# DATASET
# ==========================================

def visualizar_dataset():
    print("\n--- DATASET DE EXEMPLO ---")

    dados = {
        "Operação": [
            "Soma",
            "Multiplicação",
            "Média",
            "Área do quadrado",
            "Progressão"
        ],
        "Exemplo": [
            "10 + 5",
            "10 * 5",
            "10, 20, 30",
            "5²",
            "2, 4, 6, 8"
        ]
    }

    tabela = pd.DataFrame(dados)

    print(tabela)


# ==========================================
# MENU PRINCIPAL
# ==========================================

def menu():

    while True:

        print("\n================================")
        print("   AUTOMATIZAÇÃO MATEMÁTICA")
        print("================================")

        print("1 - Operações básicas")
        print("2 - Expressão numérica")
        print("3 - Estatística")
        print("4 - Matrizes")
        print("5 - Funções matemáticas")
        print("6 - Progressão aritmética")
        print("7 - Geometria")
        print("8 - Visualizar dataset")
        print("0 - Sair")

        opcao = input("\nEscolha uma opção: ")

        if opcao == "1":
            operacoes_basicas()

        elif opcao == "2":
            expressao_numerica()

        elif opcao == "3":
            estatistica()

        elif opcao == "4":
            matrizes()

        elif opcao == "5":
            funcoes_matematicas()

        elif opcao == "6":
            progressao_aritmetica()

        elif opcao == "7":
            geometria()

        elif opcao == "8":
            visualizar_dataset()

        elif opcao == "0":
            print("\nPrograma encerrado.")
            break

        else:
            print("\nOpção inválida.")

        input("\nPressione ENTER para continuar...")


# ==========================================
# INÍCIO DO PROGRAMA
# ==========================================

menu()
```
